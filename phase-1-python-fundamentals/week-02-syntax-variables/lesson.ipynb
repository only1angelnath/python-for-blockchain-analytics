## 2. Variables

A **variable** is a name that holds a value.

Think of it like labeling a box — you put something inside and refer to it by the label later.

**Blockchain analogy:** A wallet address is a label pointing to an account on-chain. A variable is a label pointing to a value in memory.

````python
# Storing a wallet address
wallet_address = "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045"

# Storing a token balance (in Wei — smallest ETH unit)
balance_wei = 1500000000000000000   # 1.5 ETH in Wei

# Storing a token name
token_name = "Ethereum"

# Storing whether a wallet is flagged
is_flagged = False

# Print them all
print("Address:", wallet_address)
print("Balance (Wei):", balance_wei)
print("Token:", token_name)
print("Flagged:", is_flagged)
````

### Variable naming rules

```
✅ Good names          ❌ Bad names
wallet_address         wallet address   (no spaces)
token_price_usd        2token           (can't start with a number)
is_active              class            (reserved Python word)
total_supply_eth       TotalSupply$     (no special characters except _)
```

**Convention:** Python uses `snake_case` — all lowercase, words separated by underscores. This is the standard for the entire course.

````python
# Reassigning a variable — the new value replaces the old one
gas_price = 20    # gwei
print("Gas price:", gas_price, "gwei")

gas_price = 45    # network got busy
print("Gas price:", gas_price, "gwei")

# Multiple assignment on one line
eth_price, btc_price, sol_price = 3200, 67000, 180
print(eth_price, btc_price, sol_price)
````

## 3. Data Types

Python has four core types you'll use constantly.

| Type | What it holds | Blockchain example |
|------|--------------|-------------------|
| `int` | Whole numbers | Block number, nonce, token decimals |
| `float` | Decimal numbers | Token price in USD, APY percentage |
| `str` | Text | Wallet address, tx hash, token symbol |
| `bool` | True or False | Is contract verified? Is wallet flagged? |

Use `type()` to check what type a variable is.

````python
# int — whole numbers, no decimal
block_number = 19_847_293      # underscores make big numbers readable
nonce = 42
token_decimals = 18

print(type(block_number))   # <class 'int'>

# float — decimal numbers
eth_price_usd = 3247.85
apy_percent = 4.72
slippage = 0.005

print(type(eth_price_usd))  # <class 'float'>

# str — text, always in quotes (single or double, both work)
tx_hash = "0xabc123def456..."
token_symbol = "ETH"
contract_address = '0x1f9840a85d5aF5bf1D1762F925BDADdC4201F984'

print(type(tx_hash))        # <class 'str'>

# bool — only two values: True or False (capital T and F)
is_verified = True
is_honeypot = False
has_liquidity = True

print(type(is_verified))    # <class 'bool'>
````

## 4. Numbers in Depth

### int
Integers are exact. Use them for counts, block numbers, token amounts in their smallest unit (Wei for ETH).

**Important:** Never store ETH values as floats. Floats have rounding errors that can cause financial bugs. Always work in Wei (int) and convert for display only.

````python
# Integer arithmetic
block_number = 19_847_293
blocks_per_day = 7200           # ~12 sec per block on Ethereum

days_since_genesis = block_number // blocks_per_day   # // is integer division
print("Days since genesis:", days_since_genesis)

# ETH amounts in Wei (correct way)
sender_balance_wei = 2_500_000_000_000_000_000   # 2.5 ETH
tx_value_wei =         500_000_000_000_000_000   # 0.5 ETH
gas_cost_wei =          21_000 * 20_000_000_000  # 21000 gas * 20 gwei

remaining_wei = sender_balance_wei - tx_value_wei - gas_cost_wei
print("Remaining Wei:", remaining_wei)
print("Remaining ETH:", remaining_wei / 10**18)  # convert for display only
````

````python
# Float arithmetic
eth_price = 3247.85
btc_price = 67_412.00
portfolio_eth = 2.5
portfolio_btc = 0.1

portfolio_value = (portfolio_eth * eth_price) + (portfolio_btc * btc_price)
print(f"Portfolio value: ${portfolio_value:,.2f}")

# ⚠️  Float rounding — why we avoid floats for token balances
print(0.1 + 0.2)          # Prints 0.30000000000000004 — not 0.3!
print(0.1 + 0.2 == 0.3)   # False!

# Safe comparison for floats
import math
print(math.isclose(0.1 + 0.2, 0.3))   # True ✓
````

## 5. Strings in Depth

Strings hold text. In blockchain code you'll work with strings constantly — addresses, hashes, token symbols, ABI fields.

**SQL parallel:** A string is like a `VARCHAR` column.

````python
# String basics
address = "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045"
symbol = "USDC"

# Length — how many characters
print(len(address))    # 42 — all Ethereum addresses are 42 chars

# Indexing — get one character (starts at 0)
print(address[0])      # 0
print(address[1])      # x
print(address[-1])     # last character

# Slicing — get a range of characters
prefix = address[:2]   # "0x"
checksum = address[2:] # the hex part
print("Prefix:", prefix)
print("First 8 chars:", address[:8])

# Upper / lower
print(symbol.lower())  # usdc
print(symbol.upper())  # USDC

# Check if address starts with "0x" — validation
print(address.startswith("0x"))   # True

# Replace
dirty_hash = "0xabc...def"
clean_hash = dirty_hash.replace("...", "")
print(clean_hash)
````

````python
# String concatenation (joining)
network = "Ethereum"
chain_id = "1"
label = network + " (Chain ID: " + chain_id + ")"
print(label)

# f-strings — the modern, clean way (use these always)
wallet = "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045"
balance_eth = 2.547831
tx_count = 1847

# f-string: prefix the string with f, then use {} for variables
print(f"Wallet: {wallet}")
print(f"Balance: {balance_eth:.4f} ETH")   # 4 decimal places
print(f"Transactions: {tx_count:,}")       # comma separator for thousands
print(f"Balance (USD): ${balance_eth * 3247.85:,.2f}")

# Multi-line strings (triple quotes) — useful for SQL queries or ABIs
query = """
SELECT 
    wallet_address,
    SUM(amount_usd) AS total_volume
FROM dex_trades
WHERE block_date >= '2024-01-01'
GROUP BY 1
ORDER BY 2 DESC
"""
print(query)
````

## 6. Booleans

Booleans are the result of comparisons and logic. They're the foundation of all decision-making in code (which we cover in Week 3).

**Blockchain use cases:** Is this contract verified? Is this wallet blacklisted? Has the timelock expired?

````python
# Direct boolean values
is_contract = True
is_eoa = False   # Externally Owned Account

# Comparisons produce booleans
eth_price = 3247.85
print(eth_price > 3000)    # True
print(eth_price > 5000)    # False
print(eth_price == 3247.85) # True

# Comparing strings
token = "USDC"
print(token == "USDC")     # True
print(token == "usdc")     # False — case sensitive!
print(token.lower() == "usdc")  # True — safe comparison

# Wallet validation example
address = "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045"
is_valid_length = len(address) == 42
is_valid_prefix = address.startswith("0x")
is_valid_address = is_valid_length and is_valid_prefix
print(f"Valid address: {is_valid_address}")
````

## 7. Type Casting

Converting a value from one type to another.

**Why this matters:** APIs return everything as strings. Blockchain data often comes as strings from JSON. You need to convert before doing math.

**SQL parallel:** Like `CAST(amount AS DECIMAL)` or `CAST(block_number AS INTEGER)` in SQL.

````python
# Converting API response strings to numbers
api_response = {
    "block_number": "19847293",      # string from JSON
    "value": "500000000000000000",   # Wei as string
    "gas_used": "21000",
    "gas_price": "20000000000"
}

# Cast to int for arithmetic
block_number = int(api_response["block_number"])
value_wei = int(api_response["value"])
gas_used = int(api_response["gas_used"])
gas_price = int(api_response["gas_price"])

# Now we can do math
value_eth = value_wei / 10**18
gas_cost_eth = (gas_used * gas_price) / 10**18

print(f"Block: {block_number:,}")
print(f"Value: {value_eth:.6f} ETH")
print(f"Gas cost: {gas_cost_eth:.6f} ETH")

# int → str (for building URLs or display)
base_url = "https://etherscan.io/block/"
full_url = base_url + str(block_number)
print(full_url)

# str → float
price_str = "3247.85"
price_float = float(price_str)
print(f"Price: ${price_float * 1.05:,.2f}")  # with 5% markup

# bool casting — 0 and empty string are False, everything else is True
print(bool(0))        # False
print(bool(1))        # True
print(bool(""))       # False
print(bool("0x..."))  # True
````

## 8. Putting It Together — Transaction Receipt Formatter

This is your **mini-project** for Week 2.

We'll take raw transaction data (as it comes from an API — all strings) and produce a clean, human-readable receipt.

````python
# Raw transaction data from Etherscan API (everything is a string)
raw_tx = {
    "hash": "0x5c504ed432cb51138bcf09aa5e8a410dd4a1e204ef84bfed1be16dfba1b22060",
    "from": "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045",
    "to": "0x1f9840a85d5aF5bf1D1762F925BDADdC4201F984",
    "value": "1500000000000000000",
    "gasUsed": "21000",
    "gasPrice": "20000000000",
    "blockNumber": "19847293",
    "status": "1"
}

# --- Parse and convert ---
tx_hash      = raw_tx["hash"]
sender       = raw_tx["from"]
receiver     = raw_tx["to"]
value_wei    = int(raw_tx["value"])
gas_used     = int(raw_tx["gasUsed"])
gas_price    = int(raw_tx["gasPrice"])
block_number = int(raw_tx["blockNumber"])
success      = raw_tx["status"] == "1"   # "1" means success

# --- Calculate derived values ---
value_eth    = value_wei / 10**18
gas_cost_wei = gas_used * gas_price
gas_cost_eth = gas_cost_wei / 10**18
eth_price    = 3247.85                   # pretend we fetched this from CoinGecko
value_usd    = value_eth * eth_price
gas_cost_usd = gas_cost_eth * eth_price

# --- Format and display ---
status_label = "✅ Success" if success else "❌ Failed"

print("=" * 55)
print("        ETHEREUM TRANSACTION RECEIPT")
print("=" * 55)
print(f"  Status      : {status_label}")
print(f"  Block       : {block_number:,}")
print(f"  Tx Hash     : {tx_hash[:18]}...{tx_hash[-6:]}")
print(f"  From        : {sender[:10]}...{sender[-6:]}")
print(f"  To          : {receiver[:10]}...{receiver[-6:]}")
print("-" * 55)
print(f"  Value       : {value_eth:.6f} ETH  (${value_usd:,.2f})")
print(f"  Gas Used    : {gas_used:,} units")
print(f"  Gas Price   : {gas_price // 10**9} Gwei")
print(f"  Gas Cost    : {gas_cost_eth:.8f} ETH  (${gas_cost_usd:.4f})")
print("=" * 55)
````
